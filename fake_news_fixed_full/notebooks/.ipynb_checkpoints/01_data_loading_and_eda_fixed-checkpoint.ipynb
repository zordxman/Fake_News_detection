# Fake News Detection — Clean End-to-End Pipeline

This notebook removes duplicate cells and runs the project in this order:

1. Load and clean the dataset
2. EDA
3. Train TF-IDF + LinearSVC
4. Evaluate and save the model
5. Predict a news article
6. Use Ollama for claim extraction
7. Retrieve evidence from Google News
8. Verify claims with the LLM

**Important:** the ML section works without Ollama. The LLM/evidence section requires Ollama running locally with the configured model.


```python
from pathlib import Path
import sys
import re
import string
import time
import joblib
import pandas as pd

# Resolve project root robustly whether the notebook is opened from
# notebooks/, project root, or another working directory.
CANDIDATE_ROOTS = [
    Path.cwd(),
    Path.cwd().parent,
    Path.cwd().parent.parent,
]

PROJECT_ROOT = next(
    (p for p in CANDIDATE_ROOTS
     if (p / "data" / "fake.csv").exists() and (p / "data" / "true.csv").exists()),
    Path.cwd().parent,
)

DATA_DIR = PROJECT_ROOT / "data"
MODEL_DIR = PROJECT_ROOT / "models"
SRC_DIR = PROJECT_ROOT / "src"

MODEL_DIR.mkdir(parents=True, exist_ok=True)
SRC_DIR.mkdir(parents=True, exist_ok=True)

if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

FAKE_FILE = DATA_DIR / "fake.csv"
TRUE_FILE = DATA_DIR / "true.csv"

print("Project root:", PROJECT_ROOT)
print("Fake dataset:", FAKE_FILE)
print("True dataset:", TRUE_FILE)
```

```python
# Load datasets once — no duplicate loading cells.

if not FAKE_FILE.exists():
   raise FileNotFoundError(f"Missing file: {FAKE_FILE}")

if not TRUE_FILE.exists():
    raise FileNotFoundError(f"Missing file: {TRUE_FILE}")

fake = pd.read_csv(FAKE_FILE)
true = pd.read_csv(TRUE_FILE)

required = {"text"}
missing_fake = required - set(fake.columns)
missing_true = required - set(true.columns)

if missing_fake:
    raise ValueError(f"fake.csv is missing columns: {missing_fake}")
if missing_true:
    raise ValueError(f"true.csv is missing columns: {missing_true}")

fake["label"] = 0
true["label"] = 1

df = pd.concat([fake, true], ignore_index=True)
df["text"] = df["text"].fillna("").astype(str)

# Remove empty articles and duplicate rows.
df = df[df["text"].str.strip().ne("")]
df = df.drop_duplicates().reset_index(drop=True)

# Shuffle once.
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

print("Fake news shape:", fake.shape)
print("True news shape:", true.shape)
print("Combined dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
# Dataset quality checks
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
print("\nLabel counts:")
print(df["label"].value_counts().rename(index={0: "Fake", 1: "Real"}))

if "subject" in df.columns:
    print("\nSubjects:")
    print(df["subject"].value_counts())

if "date" in df.columns:
    print("\nDate range:")
    print(df["date"].min(), "to", df["date"].max())
```

```python
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(r"<[^>]*>", " ", text)
    text = text.translate(str.maketrans("", "", string.punctuation))
    text = re.sub(r"\d+", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df["clean_text"] = df["text"].apply(clean_text)
df = df[df["clean_text"].str.strip().ne("")].reset_index(drop=True)

print(df[["text", "clean_text"]].head())
```

## Machine Learning — TF-IDF + Linear SVM

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

X = df["clean_text"]
y = df["label"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

tfidf = TfidfVectorizer(
    max_features=50000,
    ngram_range=(1, 2),
    sublinear_tf=True
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)

model = LinearSVC()
model.fit(X_train_tfidf, y_train)

y_pred = model.predict(X_test_tfidf)

accuracy = accuracy_score(y_test, y_pred)
print(f"\nAccuracy: {accuracy:.4f} ({accuracy*100:.2f}%)")
print("\nClassification Report:")
print(classification_report(
    y_test, y_pred,
    target_names=["Fake", "Real"],
    zero_division=0
))
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=["Fake", "Real"],
    yticklabels=["Fake", "Real"]
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Fake News Detection — Confusion Matrix")
plt.tight_layout()
plt.show()

TFIDF_FILE = MODEL_DIR / "tfidf_vectorizer.pkl"
MODEL_FILE = MODEL_DIR / "fake_news_svm.pkl"

joblib.dump(tfidf, TFIDF_FILE)
joblib.dump(model, MODEL_FILE)

print("Saved:", TFIDF_FILE)
print("Saved:", MODEL_FILE)
```

```python
def predict_news(news_text):
    cleaned = clean_text(news_text)
    if not cleaned:
        raise ValueError("News text is empty after cleaning.")

    vectorized = tfidf.transform([cleaned])
    prediction = int(model.predict(vectorized)[0])
    score = float(model.decision_function(vectorized)[0])

    label = "FAKE" if prediction == 0 else "REAL"
    return label, score

news = """
The government announced a new policy that will provide free
education to all students starting next year.
"""

label, score = predict_news(news)
print("Prediction:", label)
print("Decision score:", score)
```

## LLM Claim Extraction

```python
# The LLM stage is optional. The ML model above does not require Ollama.

try:
    from llm import ask_llm, extract_claims
    LLM_AVAILABLE = True
    print("LLM module imported.")
except Exception as e:
    LLM_AVAILABLE = False
    print("LLM module could not be imported:", e)
```

```python
sample_news = df.iloc[0]["text"]
print(sample_news[:1000])

if LLM_AVAILABLE:
    try:
        response = ask_llm(
            f"""Analyze this news article.

ARTICLE:
{sample_news}

Give me:
1. A short summary
2. The main factual claims
3. Important people or organizations mentioned
4. Claims that should be verified

Do not automatically say whether the article is fake or real.
"""
        )
        print(response)
    except Exception as e:
        print("Ollama request failed:", e)
else:
    print("Start Ollama and load the configured model to use this section.")
```

```python
def parse_claim_lines(raw_claims):
    """Extract CLAIM n: ... lines without assuming exactly 3 claims."""
    claims = []
    for line in str(raw_claims).splitlines():
        line = line.strip()
        if re.match(r"^CLAIM\s*\d+\s*:", line, flags=re.I):
            claim = line.split(":", 1)[1].strip()
            if claim:
                claims.append(claim)
    return claims[:3]

if LLM_AVAILABLE:
    try:
        claims_raw = extract_claims(sample_news)
        claim_lines = parse_claim_lines(claims_raw)
        print("Claims found:", len(claim_lines))
        for i, claim in enumerate(claim_lines, 1):
            print(f"CLAIM {i}: {claim}")
    except Exception as e:
        claim_lines = []
        print("Claim extraction failed:", e)
else:
    claim_lines = []
```

## Evidence Retrieval

```python
try:
    from retrieval import search_evidence
    RETRIEVAL_AVAILABLE = True
    print("Retrieval module imported.")
except Exception as e:
    RETRIEVAL_AVAILABLE = False
    print("Retrieval module could not be imported:", e)
```

```python
def get_evidence_for_claims(claims, max_records=3, delay_seconds=5):
    all_evidence = []

    if not RETRIEVAL_AVAILABLE:
        return all_evidence

    for i, claim in enumerate(claims, start=1):
        print("\n" + "=" * 80)
        print(f"SEARCHING EVIDENCE FOR CLAIM {i}")
        print(claim)

        try:
            evidence = search_evidence(claim, max_records=max_records)
            print("Evidence found:", len(evidence))
        except Exception as e:
            print("Evidence search failed:", e)
            evidence = []

        all_evidence.append({
            "claim": claim,
            "evidence": evidence
        })

        if i < len(claims) and delay_seconds:
            time.sleep(delay_seconds)

    return all_evidence

evidence_data = get_evidence_for_claims(claim_lines)
```

## Claim Verification

```python
try:
    from verification import verify_claim
    VERIFICATION_AVAILABLE = True
    print("Verification module imported.")
except Exception as e:
    VERIFICATION_AVAILABLE = False
    print("Verification module could not be imported:", e)
```

```python
verification_data = []

if VERIFICATION_AVAILABLE:
    for item in evidence_data:
        claim = item["claim"]
        evidence = item["evidence"]

        print("\n" + "=" * 80)
        print("CLAIM:")
        print(claim)

        try:
            result = verify_claim(claim, evidence)
        except Exception as e:
            result = (
                "STATUS: UNVERIFIED\n"
                f"REASON: Verification failed: {e}"
            )

        print("\nVERIFICATION:")
        print(result)

        verification_data.append({
            "claim": claim,
            "evidence": evidence,
            "verification": result
        })
else:
    print("Verification skipped because the module is unavailable.")
```

## Final Analysis Function

```python
def final_analysis(news_text):
    """Run ML prediction, claim extraction, evidence retrieval and verification."""

    print("=" * 80)
    print("FINAL FAKE NEWS ANALYSIS")
    print("=" * 80)

    # 1. ML
    svm_result, decision_score = predict_news(news_text)

    print("\n[1] ML PREDICTION")
    print("SVM Result:", svm_result)
    print("Decision score:", decision_score)

    # 2. Claims
    print("\n[2] CLAIM EXTRACTION")
    if not LLM_AVAILABLE:
        print("LLM unavailable — claim extraction skipped.")
        return {
            "svm_prediction": svm_result,
            "decision_score": decision_score,
            "claims": []
        }

    try:
        extracted = extract_claims(news_text)
        claims = parse_claim_lines(extracted)
    except Exception as e:
        print("Claim extraction failed:", e)
        claims = []

    print("Claims found:", len(claims))
    for i, claim in enumerate(claims, 1):
        print(f"{i}. {claim}")

    # 3. Evidence
    print("\n[3] EVIDENCE RETRIEVAL")
    evidence_data = get_evidence_for_claims(claims)

    # 4. Verification
    print("\n[4] CLAIM VERIFICATION")
    verification_data = []

    if VERIFICATION_AVAILABLE:
        for item in evidence_data:
            try:
                result = verify_claim(item["claim"], item["evidence"])
            except Exception as e:
                result = (
                    "STATUS: UNVERIFIED\n"
                    f"REASON: Verification failed: {e}"
                )

            verification_data.append({
                "claim": item["claim"],
                "evidence": item["evidence"],
                "verification": result
            })

            print("\nClaim:", item["claim"])
            print(result)

    return {
        "svm_prediction": svm_result,
        "decision_score": decision_score,
        "claims": verification_data
    }
```
